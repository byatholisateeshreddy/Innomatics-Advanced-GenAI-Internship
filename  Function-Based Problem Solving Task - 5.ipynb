# Problem 1: Social Media – Post Engagement Analyzer
## Problem Statement
Create a function to analyze engagement on a social media post.


```python
def engage_analyzer(likes):
    total = 0

    for i in likes:
        total += i
    if total >= 1000:
        status = "Viral Post"
    else:
        status = "Normal Engagement"

    print("Total Likes:",total)
    print("Post Status:",status)

engage_analyzer([870,730,865])
```

# Problem 3: Agriculture – Rainfall Adequacy Checker
## Problem Statement
Create a function that analyzes seasonal rainfall data.


```python
def rainfall_check(data,required):
    avg = sum(data)/len(data)

    if avg >= required:
        status = "Adequate Rainfall"
    else:
        status = "Inadequate Rainfall"

    print("Average Rainfall:",avg)
    print("Rainfall status:",status)

rainfall_check([87,73,86,80,92],80)
```

# Problem 4: Social Media – Duplicate Account Detection
## Problem Statement
Create a function to detect duplicate usernames in a system.


```python
def check_duplicates(users):
    if len(users) != len(set(users)):
        status = "Yes"
    else:
        status = "No"
    print("Duplicate Accounts Found:",status)

check_duplicates(["SAteesh","Kumar","Reddy","Satish"])
```

# Problem 6: Agriculture – Premium Crop Price Filter
## Problem Statement
Create a function that filters crop prices above a premium threshold.


```python
def crop_premium(prices):
    rate = []
    for i in prices:
        if i > 2000:
            rate.append(i)
    print("Premium Crops:",rate)

crop_premium([2500,5600,2400,1000,1300])
```

# Problem 7: System Monitoring – Application Health Checker
## Problem Statement

```python
def app_health_check(error):
    if error == 0:
        status = "Healthy"
    elif error <= 5:
        status = "Minor Issues"
    else:
        status = "Critical Issues"

    print("Error Count:",error)
    print("System Status:",status)

app_health_check(5)
```

# Problem 10: Smart Electricity Bill Analyzer
## Problem Statement
You are building a smart electricity billing system for residential users.
Create a function that calculates the electricity bill and classifies usage level based on units consumed.


```python
def electricity_bill(units):
    bill = 0

    if units <= 100:
        bill = units * 3
    elif units <= 200:
        bill = (100*3) + (units-100)*5
    else:
        bill = (100*3) + (100*5) + (units-200)*7

    if bill < 500:
        status = "Low Usage"
    elif bill <= 1500:
        status = "Moderate Usage"
    else:
        status = "High Usage"
    return bill, status
    
amount,level = electricity_bill(150)
print("Bill:",amount)
print("Usage:",level)
```
