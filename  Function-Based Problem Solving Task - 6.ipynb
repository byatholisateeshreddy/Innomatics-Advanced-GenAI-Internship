# Problem Statement 1:Smart Parking Lot Management System
### Design a function to manage a smart parking lot.
- The system should:
- Accept vehicle entry and exit logs
- Calculate total parked vehicles
- Identify peak parking usage
- Alert if parking exceeds capacity

```python
def smart_parking(capacity, logs):
    parked = 0

    for i in logs:
        i = i.upper()
        if i == "IN":
            parked += 1
        elif i == "OUT":
            parked -= 1
    print("Currently Parked Vehicles:",parked)

    if parked >= capacity:
        status = "Exceeded Capacity"
    else:
        status = "Available"
    print("Parking Status:",status)

smart_parking(50, ["IN", "IN", "IN", "OUT", "IN", "IN", "OUT"])
```

# Problem Statement 2:Online Food Delivery Time Estimator
### Create a function that estimates delivery time based on:
- Distance (km)
- Weather condition
- Traffic level
- Apply delays dynamically and display final ETA.

```python
def food_delivery_time(distance, traffic, weather):
    time = distance * 4
    traffic = traffic.capitalize()
    weather = traffic.capitalize()

    if traffic == "High":
        time += 20
    elif traffic == "Medium":
        time += 15
    else:
        time += 10

    if weather == "Rainy":
        time += 8
    elif weather == "Strom":
        time += 15

    print("Estimated Delivery Time:",time,"minutes")

food_delivery_time(8, "High", "Rainy")
```

# Problem Statement 3:Movie Theatre Seat Occupancy Analyzer
- Build a function that analyzes seat booking data and:
- Calculates occupancy percentage
- Determines if show is Housefull
- Suggests opening additional shows

```python
def theatre_seat_occupancy(total_seats, booked_seats):
    count = len(booked_seats)
    occupancy = (count/total_seats) * 100

    if occupancy == 100:
        status = "Housefull"
    elif 75 <= occupancy < 100:
        status = "Almost Full"
    else:
        status = "Seats Available"
        
    print(f"Occupancy: {int(occupancy)}%")
    print("Show Status:",status)

theatre_seat_occupancy(200, [1]*150)
```

# Problem Statement 4:Cloud Server Load Classification System
- Create a function to classify server load based on CPU usage readings.
- Rules:
- Average CPU < 50% → Normal
- 50%–80% → Warning
- 80% → Critical

```python
def cloud_server_load(cpu_readings):
    total = 0

    for i in cpu_readings:
        total += i
    average = total/len(cpu_readings)

    if average < 50:
        status = "Normal"
    elif 50 <= average <= 80:
        status = "Warning"
    else:
        status = "Critical"

    print(f"Average CPU Load: {int(average)}%")
    print("Server Status:",status)

cloud_server_load([45, 60, 70, 85, 90])
```

# Problem Statement 5:Smart Classroom Resource Usage Monitor
- Design a function that tracks usage of classroom resources (projector, AC, lights) and identifies overuse patterns.


```python
def smart_classroom_resource(usage):
    overused = []

    for i in usage:
        if usage[i] > 8:
            overused.append(i)

    if overused:
        print("Overused Resource:",",".join(overused))
        print("Energy Alert: Yes")
    else:
        print("Overused Resource: None")
        print("Energy Alert: No")

smart_classroom_resource({"Projector":6, "AC":9, "Lights":4})
```

# Problem Statement 6: Online Event Registration Capacity Controller
 - Create a function that manages event registrations by:
1. Tracking registrations
2. Preventing overbooking
3. Triggering waitlist mode


```python
def online_event_registration(capacity, registers):
    if registers > capacity:
        confirmed = capacity
        waitlist = registers - capacity
        status = "Closed"
    else:
        confirmed = registers
        waitlist = 0
        status = "Open"

    print("Confirmed Registrations:",confirmed)
    print("Waitlisted Users:",waitlist)
    print("Registrations status:",status)

online_event_registration(100, 105)
```
