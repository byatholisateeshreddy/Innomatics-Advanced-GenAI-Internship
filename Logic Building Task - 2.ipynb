# Problem 1: Unique Words in a Sentence
### Problem Statement
Given a sentence, identify all unique words and display:
1. The count of unique words
2. The unique words themselves

```python
sentence = "python is easy and python is powerful"
words = sentence.split()  # splitting the sentence into indiviual words using split() function
set_unique = set(words)  # removing duplicates using set

print("Unique words count: ",len(set_unique))
print("Unique words:",set_unique)
```

# Problem 2: Highest Salary from Employee Data
### Problem Statement
Store employee names and their salaries, and determine which employee has the highest salary

```python
employees = {"Ravi": 75000,"Anita": 68000,"Kiran": 72000}
highest_salary = 0
highest_earner = ""
for name,salary in employees.items():
    if salary >= highest_salary:
        highest_salary = salary
        highest_earner = name
        
print("Highest Salary:", highest_earner, "-", highest_salary)
```

# Problem 3: Find Maximum and Minimum Values
### Problem Statement
Find the maximum and minimum values from a list of numbers.

```python
numbers = [45,22,89,10,66]
maximum = numbers[0]
minimum = numbers[0]

for i in numbers:
    if i > maximum:
        maximum = i
    if i < minimum:
        minimum = i
        
print("List:", numbers)
print("Max:", maximum)
print("Min:", minimum)
```

# Problem 4: Count Products Above a Price Threshold
### Problem Statement
Count how many products have a price greater than 1000.

```python
price = [450, 1200, 899, 1500, 300]
count = 0
for i in price:
    if i > 1000:
        count = count+1
print("Products above 1000:", count)
```

# Problem 5: Calculate Attendance Percentage
### Problem Statement
Calculate the attendance percentage from a list containing "P" (Present) and "A" (Absent).

```python
attendance = ["P", "P", "A", "P", "P"]
count = 0
for i in attendance:
    if i == "P":
        count = count + 1
result = (count/len(attendance)*100)
print("Attendance Percenatge:", result)
```

# Problem 10: Calculate Average Marks
### Problem Statement
Calculate the average marks of a student from a list of marks.
Hint
Calculate the sum of all values.
Divide by the total number of marks.
Expected Output
Average Marks: 72.5

Real-Time Applications
Student performance evaluation
Result processing systems


```python
marks = [87,73,86,65,69,55]
total = 0

for i in marks:
    total = total + i
average = total/len(marks)
    
print("Average MArks:",average)
```
