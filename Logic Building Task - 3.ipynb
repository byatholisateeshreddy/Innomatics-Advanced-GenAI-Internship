# Problem Statement 1: Employee Performance Bonus Eligibility

```python
employees = {
"Ravi": 92,
"Anita": 88,
"Kiran": 92,
"Suresh": 85
}
# finding highest score from employees
highest_score = max(employees.values())
# identifing name of employees with the highest score
top_performance = []
for name, score in employees.items():
    if score == highest_score:
        top_performance.append(name)

print("Top Performers Eligible for Bonus:", ", ".join(top_performance), "(Score:",highest_score,")")
```

# Problem Statement 2: Search Query Keyword Analysis

```python
sentence = "Buy mobile phone buy phone online"
# converting input to lowercase
lower_case = sentence.lower()
# splitting the sentence 
words = lower_case.split()
# counting frequency of each words
word_count = {}
for i in words:
    if i in word_count:
        word_count[i]+=1
    else:
        word_count[i]=1
# finding the word occuring more than one time
keywords = {}
for i, count in word_count.items():
    if count >= 2:
        keywords[i] = count

print(keywords)
```

# Problem Statement 3: Sensor Data Validation

```python
sensor_readings = [3, 4, 7, 8, 10, 12, 5]
# empty list to store readings
valid_readings = []

for i in range(len(sensor_readings)): 
    j = sensor_readings[i]
    if j%2 == 0:
        valid_readings.append((i, j))
print(valid_readings)
```

# Problem Statement 4: Email Domain Usage Analysis

```python
emails = [
"ravi@gmail.com",
"anita@yahoo.com",
"kiran@gmail.com",
"suresh@gmail.com",
"meena@yahoo.com"
]
domain_count = {}
for i in emails:
    domain = i.split("@") [1]
    if domain in domain_count:
        domain_count[domain] += 1
    else:
        domain_count[domain] = 1

for i, j in domain_count.items():
    print(f"{i}: {(j/len(emails))*100:.0f}%")
```

# Problem Statement 5: Sales Spike Detection

```python
sales = [1200, 1500, 900, 2200, 1400, 3000]
avg_sales = sum(sales)/len(sales)   # calculating the avg 
threshold = avg_sales + (avg_sales * 0.3)  # more than 30% above avg

day_number = 1
for sale_value in sales:
    if sale_value > threshold:
        print(f"Day {day_number}: {sale_value}")
    day_number = day_number + 1
```

# Problem Statement 6: Duplicate User ID Detection

```python
user_ids = ["user1", "user2", "user1", "user3", "user1", "user3"]

count_ids = {}
for i in user_ids:
    if i in count_ids:
        count_ids[i] += 1
    else:
        count_ids[i] = 1
        
for k, j in count_ids.items():
    if j >= 2:
        print(k,"->",j,"times")
```
