# Problem 1: Mobile Recharge Validation System
# Problem Statement
Create a function to validate a mobile recharge request.
## Rules:
Recharge amount must be ≥ ₹50

Valid plans: 199, 299, 399, 599

User should be allowed to retry if invalid

```python
def validate_recharge():
    valid_plans = [199, 299, 399, 599]

    while True:
        amount = int(input("Enter recharge amount:"))

        if amount <= 50:
            print("Recharge must be greater than or equal to ₹50")
        elif amount not in valid_plans:
            print("Invalid: This Plan not available")
        else:
            print("Recharge Successful!")
            break

validate_recharge()
```

# Problem 2: Inventory Reorder Alert System
# Problem Statement
## Given a dictionary of products and stock quantities:
Print Reorder Alert if stock < 15

Print Stock OK otherwise

```python
def stock_quantity(products):
    for item, quantity in products.items():
        if quantity < 15:
            print(item, "---> Recoder Alert\n")
        else:
            print(item, "---> Stock Ok\n")
            
stock_quantity({"Laptop":10, "Tab":5, "Desktop":8, "Phone":20})
```

# Problem 3: Student Result Processing System
# Problem Statement
## Create a function that:
Accepts a list of student marks

Calculates average marks

Prints Pass if average ≥ 50 else Fail


```python
def student_grade(marks):
    # finding the avg of the student
    average = sum(marks)/len(marks)
    # checking student is pass or fail
    if average >= 50:
        print("The student is pass")
    else:
        print("The student is fail")
        
student_grade([87,73,86,90,95,80])
```

# Problem 4: Cab Fare Estimator with Retry Option
# Problem Statement
## Create a cab fare calculator:
Base fare = ₹50

₹12 per km

Peak hour → 25% extra

Allow retry using while loop


```python
# creating the function with no paramters
def cab_fare():
    distance = int(input("Enter the distance in KM:"))
    peak = input("Is It peak hour? (yes/no):").lower()

    # calculating the fare amount
    fare_amount = 50 + (distance * 12)
    while True:
        # At peak hour
        if peak == "yes":
            total_amount = fare_amount + (fare_amount * 0.25)
            print(f"Total Fare = {total_amount}Rs.")
        else:
            print(f"Total Fare = {total_amount}Rs.")
       # Asking the customer about booking
        asking = input("If you want to book cab? (yes/no):").lower()
       # booking connfirm
        if asking == "yes":
            print("Booking confirmed!")
            print("Thank you for booking and have a safe ride")
            break
        else:
            print("Booking cancelled!")

        # asking to retry 
        retry = input("You want to retry? (yes/no):")
        if retry == "no":
            break
            
cab_fare()                   
```

# Problem 5: Employee Attendance Eligibility Checker
# Problem Statement
## Create a function that:
Accepts attendance list ("P", "A")

Calculates attendance percentage

Returns Eligible if ≥ 75% else Not Eligible


```python
# creating function with one parameter
def students_eligibility(Attendance):
    
    # couting how many presents
    present_count = 0
    for present in Attendance:
        
        # converting into upper case letter
        present = present.upper()
        if present == "P":
            present_count += 1
            
    # calculating the percentage of present
    percentage = (present_count / len(Attendance)) * 100
    
    # checking eligibility
    if percentage >= 75:
        return "Eligible"
    else:
        return "Not Eligible"
        
students_eligibility(["P", "P", "P", "A", "A", "P", "P", "P", "P", "P"])
```

# Problem 6: Password Strength Checker
# Problem Statement
Create a function to check password strength.
## Rules:
Minimum length = 8

Must contain at least one digit

Must contain at least one special character (@#$)


```python
def password_checker():
    password = input("Enter your password:")
    upper = 0
    lower = 0
    digit = 0
    special = 0

    for i in password:
        if i.isupper():
            upper += 1
        if i.islower():
            lower += 1
        if i.isdigit():
            digit += 1
        if i in "@#$":
            special += 1
    if len(password) >= 8 and upper >= 1 and lower >= 1 and digit >= 1 and special >= 1:
        return "Strong Password"
    else:
        return "Weak Password"
password_checker()
```
